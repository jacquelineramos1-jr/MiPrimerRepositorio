3. $\textbf{Coeficientes binomiales.}$ \
El coeficiente binomial $\begin{pmatrix} n \\ 0 \end{pmatrix}$ es un número entero igual a: 

\begin {equation}
    \begin{pmatrix} n \\ 0 \end{pmatrix}=\frac{n!}{k!(n-k)!}
\end{equation}

donde k$\geq$1, o bien $\begin{pmatrix} n \\ 0 \end{pmatrix} = 1$ cuando $k=0$.

c) La probabilidad de que para una moneda no sesgada, lanzada $n$ veces, salga águila $k$ veces es: 

\begin{equation}
    p(k|n)=\frac{\begin{pmatrix} n \\ 0 \end{pmatrix}}{2^n}
\end{equation}

Escribe un programa para calcular:

1) La probabilidad total de ue una moneda lanzada 100 veces, salga águila exactamente 60 veces.
2) La probabilidad de que salga águila 60 veces o más.

```python
a) 
def binomial(n,k):
    #Caso especial: k = 0
    if k == 0: 
        return 1
    producto = 1
    for i in range(1, k + 1):
        producto = producto * (n-i+1) // i
    return producto
    
b) 
print("Triángulo de Pascal")
print("---------")

for n in range(20):
    for k in range(n+1):
        print(binomial(n,k), end=" ")
    print()

c) #Caso1: Exactamente 60 águilas
n = 100
k = 60

p_60 = binomial(n, k) / 2**n

print("\nPROBABILIDADES")
print("---------")

print("Probabilidad de obtener exactamente 60 águilas:")
print(p_60)

#Caso2: 60 águilas o más
probabilidad_60_o_mas = 0 
for k in range(60, 101):
    probabilidad_60_o_mas += binomial(n, k) / 2**n
    
print("\nProbabilidad de obtener 60 águilas o más:")
print(probabilidad_60_o_mas)
```
